````python
import numpy as np
import csv
# 1
file_list = ['mars_base_main_parts-001.csv','mars_base_main_parts-002.csv','mars_base_main_parts-003.csv']

arrays = []
for file in file_list:
  arr=np.loadtxt(file,dtype='str', delimiter=',', encoding='utf-8', skiprows=1)
  arrays.append(arr)

parts = arrays[0][:,0] 
strength1=arrays[0][:,1].astype(int)
strength2=arrays[1][:,1].astype(int)
strength3=arrays[2][:,1].astype(int)

# 열별 모아서 배열 안에 저장해줌
strengths = np.column_stack([
    strength1,
    strength2,
    strength3
])

strengths_mean =np.round(np.mean(strengths, axis=1),3) #각 행별 평균
# 3-1 parts_to_work_on.csv
try:
 with open("parts_to_work_on.csv", mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    for i in range(100):
     mean=strengths_mean[i] 
     # print(f'평균 : {mean}')
     if 50.0 > (mean):
      writer.writerow([parts[i],mean])
    print("저장 완료")

except OSError as e:
    print(f"파일 저장 중 오류가 발생했습니다: {e}")
````

`csv.writer`는 “한 줄(row)”을 써야 해서, 보통 리스트나 튜플처럼 여러 값을 담은 형태를 받습니다.

`mean`은 그냥 하나의 숫자라서:

```python
writer.writerow(mean)
```

이렇게 쓰면 안 됩니다.  
왜냐하면 `csv`가 `mean`을 “여러 칼럼이 들어 있는 순서형 데이터”로 해석하려고 하기 때문인데, 숫자 하나는 그걸 반복할 수 없어서 에러가 납니다.

맞는 방식은:

```python
writer.writerow([mean])      # mean 1개만 쓰기
```

또는 두 개를 같이 쓰려면:

```python
writer.writerow([parts[i], mean])
```

즉:
- 1개 값 쓰기 → `writer.writerow([mean])`
- 2개 값 쓰기 → `writer.writerow([parts[i], mean])`

원하면 소수점 3자리로 저장하려면:

```python
writer.writerow([parts[i], f"{mean:.3f}"])
```

print() 결과가 [4 5]처럼 쉼표 없이 괄호 안에 공백으로 구분되어 있다면, 그것이 바로 NumPy 배열이라는 뜻입니다!

np.array() 배열 생성 방법
 "뷰(View)를 반환한다"는 말은 슬라이싱을 한 순간 바로 원본이 바뀌는 것이 아니라, "원본과 같은 메모리를 바라보는 (창문 같은) 참조를 만든다"는 뜻입니다.

---

## **1. 왜 `arr1[:3]`을 프린트했을 때 원본이 그대로일까요?**

`arr1[:3]` 기호 자체는 "arr1의 앞 3개 요소가 있는 위치를 가리키는 창문(View)"을 새로 만들어서 보여줄 뿐입니다.

그 창문을 통해 **안에 있는 값을 직접 수정했을 때** 원본까지 같이 바뀌게 됩니다.

---

## **2. 파이썬 `list` vs NumPy `ndarray` 비교**

### **파이썬 `list` (복사본 생성)**

```python
py_list = [10, 20, 30, 40, 50]
sub_list = py_list[:3] # 새로운 리스트가 만들어짐 (독립된 복사본)

sub_list[0] = 999     # 슬라이싱한 리스트의 값을 변경

print(sub_list) # [999, 20, 30]
print(py_list)  # [10, 20, 30, 40, 50] (원본은 전혀 영향 없음!)

```

### **NumPy `ndarray` (뷰 생성)**

```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50])
view_arr = arr1[:3]  # arr1의 일부를 바라보는 '뷰(View)' 생성

view_arr[0] = 999    # 뷰를 통해서 값을 변경!

print(view_arr) # [999, 20, 30]
print(arr1)     # [999, 20, 30, 40, 50] (★ 원본도 같이 999로 변경됨!)

```

## **3. 요약하자면**

* **뷰(View)**: 원본 데이터가 저장된 메모리 주소를 그대로 공유합니다. 데이터 복사 비용이 들지 않아 속도가 매우 빠르지만, **슬라이싱한 변수의 값을 바꾸면 원본도 함께 바뀝니다.**
* **원본에 영향을 주고 싶지 않다면?**: `.copy()`를 사용해 완전히 새로운 배열로 복사해야 합니다.

```python
# 원본 보호하기
sub_arr = arr1[:3].copy() # 완전히 새로운 복사본을 만듦
sub_arr[0] = 777

print(sub_arr) # [777, 20, 30]
print(arr1)    # [10, 20, 30, 40, 50] (원본 안 바뀜)

```
